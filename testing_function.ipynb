```python
import pandas as pd
import os

def create_truncated_csv(original_csv_path, output_csv_path, num_rows_to_keep=5):
    """
    Reads the original CSV, keeps the header and the first 'num_rows_to_keep'
    data rows, and writes them to a new CSV file.

    Args:
        original_csv_path (str): The path to the original CSV file.
        output_csv_path (str): The path where the truncated CSV will be saved.
        num_rows_to_keep (int): The number of data rows (excluding header) to keep.
    """
    try:
        # Read only the header and the specified number of rows
        # header=0 reads the first row as header
        # nrows reads the data rows after the header
        df = pd.read_csv(original_csv_path, nrows=num_rows_to_keep)

        # Save the truncated DataFrame to a new CSV file
        df.to_csv(output_csv_path, index=False)

        print(f"Successfully created truncated CSV: '{output_csv_path}'")
        print(f"It contains the header and the first {num_rows_to_keep} data rows.")

    except FileNotFoundError:
        print(f"Error: The file '{original_csv_path}' was not found.")
    except Exception as e:
        print(f"An error occurred: {e}")

# --- Configuration ---
original_csv_file = "2008_09_15_01.csv"
truncated_csv_file = "2008_09_15_01_test.csv"
lines_to_keep = 5 # This will keep the header + 5 data rows

# --- Run the function ---
if __name__ == "__main__":
    # Create a dummy CSV file for demonstration if it doesn't exist
    # In your actual use, ensure '2008_09_15_01.csv' exists in the same directory
    if not os.path.exists(original_csv_file):
        print(f"'{original_csv_file}' not found. Creating a dummy file for demonstration.")
        dummy_data = {
            'col1': [f'data{i}' for i in range(1, 11)],
            'col2': [i * 10 for i in range(1, 11)],
            'col3': [f'category{i%3}' for i in range(1, 11)]
        }
        dummy_df = pd.DataFrame(dummy_data)
        dummy_df.to_csv(original_csv_file, index=False)
        print(f"Dummy '{original_csv_file}' created.")

    create_truncated_csv(original_csv_file, truncated_csv_file, lines_to_keep)

```
